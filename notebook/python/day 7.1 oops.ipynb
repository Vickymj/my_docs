## 1. The `super()` Function
`super()` is used within a child class to call methods or constructors from a parent class. This helps in extending the functionality of inherited methods without having to rewrite code.

### Example:

```python
# 1. `super()` Function Example
class Shape:
    def __init__(self, color):
        self.color = color
    
    def describe(self):
        print(f"This is a {self.color} shape.")

class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)  # Call to the parent class constructor
        self.radius = radius
    
    def describe(self):
        super().describe()  # Extend functionality of parent method
        print(f"It is a circle with a radius of {self.radius}.")

# Creating a Circle object
circle = Circle('red', 5)

# Using the describe method
circle.describe()
```

## 2. Abstract Classes
An **abstract class** in Python is a class that cannot be instantiated directly. It often contains abstract methods that must be implemented by any subclass that inherits from the abstract class. Python's `abc` (Abstract Base Classes) module is used to create abstract classes.

### Example:

```python
from abc import ABC, abstractmethod

# 2. Abstract Class Example
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass  # This is an abstract method

    @abstractmethod
    def perimeter(self):
        pass  # This is an abstract method

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)

# Creating a Rectangle object
rect = Rectangle(10, 5)
print(f"Area of rectangle: {rect.area()}")
print(f"Perimeter of rectangle: {rect.perimeter()}")
```

## 3. Polymorphism
Polymorphism allows methods in different classes to have the same name but act differently based on the object's class. Polymorphism in Python can be achieved in two ways:
- **Method Overriding** (inherited methods behave differently in child classes)
- **Duck Typing** (objects can be used interchangeably if they implement the required methods)

### 3.1 Method Overriding Example:

```python
# 3.1 Method Overriding Example
class Animal:
    def sound(self):
        print("This animal makes a sound.")

class Dog(Animal):
    def sound(self):
        print("The dog barks.")

class Cat(Animal):
    def sound(self):
        print("The cat meows.")

# Polymorphism in action
def make_sound(animal):
    animal.sound()

dog = Dog()
cat = Cat()
make_sound(dog)  # Output: The dog barks.
make_sound(cat)  # Output: The cat meows.
```

### 3.2 Duck Typing Example
In **Duck Typing**, Python doesn't care about the class of an object, as long as the object implements the expected behavior (methods). If an object 'quacks' like a duck and 'walks' like a duck, Python treats it as a duck.

### Example:

```python
# 3.2 Duck Typing Example
class Bird:
    def fly(self):
        print("The bird is flying.")

class Airplane:
    def fly(self):
        print("The airplane is flying.")

# Using duck typing
def perform_fly_action(obj):
    obj.fly()

bird = Bird()
plane = Airplane()
perform_fly_action(bird)   # Output: The bird is flying.
perform_fly_action(plane)  # Output: The airplane is flying.
```

## 4. Inheritance
Inheritance allows one class (child) to inherit methods and properties from another class (parent). It promotes code reuse and flexibility.

We have already seen examples of inheritance in sections above (e.g., `Animal` and `Dog`). Below is another example demonstrating inheritance.

### Example:

```python
# 4. Inheritance Example
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model
    
    def start(self):
        print(f"Starting {self.make} {self.model}.")

class Car(Vehicle):
    def __init__(self, make, model, doors):
        super().__init__(make, model)
        self.doors = doors
    
    def describe(self):
        print(f"{self.make} {self.model} with {self.doors} doors.")

# Creating a Car object
car = Car('Toyota', 'Camry', 4)
car.start()        # Inherited method
car.describe()     # Child class method
```

## Summary
- The `super()` function is used to call methods from a parent class.
- Abstract classes define methods that must be implemented in child classes.
- Polymorphism allows the same method to behave differently based on the object's class, achieved via method overriding and duck typing.
- Inheritance allows one class to inherit attributes and methods from another class, promoting code reuse and flexibility.
