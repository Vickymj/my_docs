# Python OOPs Basics: 

## 1. Introduction to Object-Oriented Programming (OOP)

- **Object-Oriented Programming (OOP)** is a programming paradigm based on objects that can contain data (attributes) and code (methods).
- OOP enables better code organization, reusability, and modularity.

## 2. Classes and Objects

### What is a Class?
- A **class** is a blueprint for creating objects (instances).
- It defines attributes (data) and methods (functions) that all objects created from the class will have.

```python
# Example of a simple class definition in Python
class Car:
    pass  # Empty class
```

### What is an Object (Instance)?
- An **object** is an instance of a class.
- Each object can have unique values for its attributes, though it shares the same methods as other instances.

```python
# Creating an instance (object) of the Car class
my_car = Car()
print(type(my_car))
```

## 3. Class Constructor (`__init__` Method)
- The `__init__` method is a **constructor**. It is automatically called when an object is created.
- It is used to initialize the attributes of the class.

```python
# Example with a constructor
class Car:
    def __init__(self, make, model, year):
        self.make = make  # Instance variable
        self.model = model  # Instance variable
        self.year = year  # Instance variable

# Creating an instance of the Car class with attributes
my_car = Car("Toyota", "Corolla", 2020)
print(my_car.make)
print(my_car.model)
```

## 4. Instance Variables
- **Instance variables** are variables that belong to each instance of the class.
- Every object (instance) has its own copy of the instance variables.

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make  # Instance variable
        self.model = model  # Instance variable
        self.year = year  # Instance variable

# Creating instances with different values for instance variables
car1 = Car("Honda", "Civic", 2021)
car2 = Car("Ford", "Focus", 2018)

print(car1.make)
print(car2.make)
```

## 5. Class Variables
- **Class variables** are shared across all instances of a class.
- All objects created from the class will access the same value of a class variable.

```python
class Car:
    wheels = 4  # Class variable, shared by all instances

    def __init__(self, make, model, year):
        self.make = make  # Instance variable
        self.model = model  # Instance variable
        self.year = year  # Instance variable

# Creating instances
car1 = Car("Honda", "Civic", 2021)
car2 = Car("Ford", "Focus", 2018)

# Accessing class variable
car1.wheels=2
print(car1.wheels)  # Output: 4

print(car2.wheels)  # Output: 4

# Changing the class variable
Car.wheels = 3
print(car1.wheels)  # Output: 3
print(car2.wheels)  # Output: 3
```

## 6. Instance Methods 
- Instance methods require an instance to be called and can access instance variables using `self`.

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    # Instance method
    def car_details(self):
        return f"{self.year} {self.make} {self.model}"

my_car = Car("Tesla", "Model S", 2022)
print(my_car.car_details())
```
