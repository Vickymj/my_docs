# Python Conditions and If Statements

Python provides several ways to perform conditional operations, which allow you to execute certain blocks of code based on specific conditions.

### Common Comparison Operators:
* **Equals (`==`)**: Returns `True` if both operands are equal.
* **Not Equals (`!=`)**: Returns `True` if operands are not equal.
* **Less than (`<`)**: Returns `True` if the left operand is less than the right operand.
* **Less than or equal to (`<=`)**: Returns `True` if the left operand is less than or equal to the right operand.
* **Greater than (`>`)**: Returns `True` if the left operand is greater than the right operand.
* **Greater than or equal to (`>=`)**: Returns `True` if the left operand is greater than or equal to the right operand.

### Basic `if` Statement
The `if` statement evaluates a condition and executes a block of code if the condition is `True`.

```python
x = 10
y = 5

if x > y:
    print("x is greater than y")
```

### Short Hand If

You can write a simple `if` statement in one line if there's only a single action to perform.

```python
a = 3
b = 4

if a < b: print("a is less than b")
```

### Short Hand If ... Else

A one-liner for `if...else` can also be used when you need to quickly evaluate conditions.

```python
a = 5
b = 2

print("a is greater") if a > b else print("b is greater")
```

### `if` with Logical Operators

You can combine multiple conditions using logical operators like `and`, `or`, and `not`.

```python
a = 10
b = 20
c = 15

if a < b and b > c:
    print("Both conditions are True")

if a < b or b < c:
    print("At least one condition is True")
```

### Nested `if` Statements

You can use one `if` statement inside another `if` statement. This is known as nested `if`.

```python
x = 10

if x > 5:
    print("x is greater than 5")
    if x > 8:
        print("x is also greater than 8")
    else:
        print("x is not greater than 8")
```

### The `pass` Statement

The `pass` statement is a null operation. It's a placeholder that does nothing, used when a statement is required syntactically but no action is needed.

```python
x = 10

if x > 5:
    pass  # Placeholder for future code

print("This will still print.")
```

# Python Loops

Loops allow you to repeatedly execute a block of code as long as a condition is met.

## `while` Loop

The `while` loop executes a set of statements as long as a condition is `True`.

```python
i = 1

while i <= 5:
    print(f"i is {i}")
    i += 1  # Increment i
```

## `for` Loop

The `for` loop in Python is used to iterate over a sequence (like a list, tuple, dictionary, set, or string) and perform an operation on each element.

```python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

### Looping Through a String

You can loop through the characters of a string using a `for` loop.

```python
for char in "banana":
    print(char)
```

### The `break` Statement

The `break` statement can be used to stop a loop before it has looped through all items.

```python
for fruit in fruits:
    if fruit == "banana":
        break
    print(fruit)
```

### The `continue` Statement

The `continue` statement skips the rest of the code inside a loop for the current iteration and moves to the next iteration.

```python
for fruit in fruits:
    if fruit == "banana":
        continue
    print(fruit)
```

### The `else` Statement in Loops

The `else` block in a loop is executed when the loop has exhausted all items (in a `for` loop) or when the condition becomes false (in a `while` loop).

```python
for i in range(5):
    print(i)
else:
    print("Loop is done")
```

### Nested Loops

A loop inside another loop is called a nested loop. The inner loop is executed one time for each iteration of the outer loop.

```python
adj = ["red", "big", "tasty"]

for color in adj:
    for fruit in fruits:
        print(color, fruit)
```

## `range()` Function

The `range()` function generates a sequence of numbers, which can be used in loops.

```python
for i in range(10):
    print(i)
```

### Specifying Start, End, and Step in `range()`

You can specify the start, end, and step values in the `range()` function.

```python
for i in range(1, 10, 2):
    print(i)
```

# Dictionary Access and Looping

### Accessing Dictionary Items

You can access the value of a specific key in a dictionary using `dict[key]`.

```python
my_dict = {"name": "Alice", "age": 25, "city": "New York"}

print(my_dict["name"])  # Output: Alice
```

### Looping Through a Dictionary

You can loop through a dictionary's keys and values.

```python
for key, value in my_dict.items():
    print(key, "->", value)
```

### Using `enumerate()` with Dictionaries

The `enumerate()` function adds a counter to an iterable, which can be useful when you need the index alongside the key-value pairs.

```python
for i, (key, value) in enumerate(my_dict.items()):
    print(i, key, "->", value)
```
