
# Python Collections (Arrays)
* **List** is a collection which is ordered and changeable. Allows duplicate members.
* **Tuple** is a collection which is ordered and unchangeable. Allows duplicate members.
* **Set** is a collection which is unordered, unchangeable*, and unindexed. No duplicate members.
*  **Dictionary** is a collection which is ordered** and changeable. No duplicate members.

###              **List**
* Lists are used to store multiple items in a single variable.

# Creating a List


```python
# Step 1: Create a list using the list() constructor
original_list = list(range(1, 6))  # Creates a list with elements [1, 2, 3, 4, 5]

# Step 2: Square each element using a list comprehension
squared_list = [1,2,3,4,5,6]

print("Original List:", original_list)
print("Squared List:", squared_list)

```

# Accessing List Items
* You can access items in a list by referring to their index.

```python
# Creating a list
my_list = ['apple', 'banana', 'cherry', 'date', 'elderberry']

# Accessing items by index
first_item = my_list[0]         # 'apple' (first item)
second_item = my_list[1]        # 'banana' (second item)
last_item = my_list[-1]         # 'elderberry' (last item using negative index)

# Modifying an item
my_list[2] = 'citrus'           # Changing 'cherry' to 'citrus'

# Accessing a slice of the list
subset = my_list[1:4]           # Slicing from index 1 to 3 (excluding 4), resulting in ['banana', 'citrus', 'date']

# Print all results
print("First item:", first_item)
print("Second item:", second_item)
print("Last item:", last_item)
print("Modified list:", my_list)
print("Subset of the list:", subset)
```

# Modifying List Items
* Change an item in the list by referring to its index and assigning a new value.

```python
# Define a list
my_list = ['apple', 'banana', 'cherry', 'date', 'elderberry']

# Print the original list
print("Original List:", my_list)

# Modify the item at index 1 (second item)
my_list[1] = 'blueberry'  # Change 'banana' to 'blueberry'

# Modify the item at index 3 (fourth item)
my_list[3] = 'dragonfruit'  # Change 'date' to 'dragonfruit'

# Print the modified list
print("Modified List:", my_list)
```

# Adding Items to a List
* Using append(): Adds an item to the end of the list.
* Using insert(): Inserts an item at a specified position.
* Using extend(): Adds items from another list to the end of the current list.

```python
# Define an initial list
my_list = ['apple', 'banana', 'cherry']

# Using append() to add an item to the end of the list
my_list.append('date')
print("After append:", my_list)  # Output: ['apple', 'banana', 'cherry', 'date']

# Using insert() to add an item at a specified position
my_list.insert(1, 'blueberry')  # Insert 'blueberry' at index 1
print("After insert:", my_list)  # Output: ['apple', 'blueberry', 'banana', 'cherry', 'date']

# Using extend() to add items from another list
another_list = ['fig', 'grape']
my_list.extend(another_list)
print("After extend:", my_list)  # Output: ['apple', 'blueberry', 'banana', 'cherry', 'date', 'fig', 'grape']
```

# Removing Items from a List
* Using remove(): Removes the first occurrence of a specified item.
* Using pop(index): Removes the item at the specified index (or the last * item if no index is specified) and returns it.
* Using clear(): Removes all items from the list.

```python
# Define an initial list
my_list = ['apple', 'banana', 'cherry', 'date', 'elderberry']

# Using remove() to remove the first occurrence of a specified item
my_list.remove('banana')
print("After remove:", my_list)  # Output: ['apple', 'cherry', 'date', 'elderberry']

# Using pop(index) to remove the item at a specified index or the last item if no index is specified
removed_item = my_list.pop(2)  # Removes item at index 2 ('date')
print("Removed item:", removed_item)  # Output: date
print("After pop:", my_list)  # Output: ['apple', 'cherry', 'elderberry']

# Using clear() to remove all items from the list
my_list.clear()
print("After clear:", my_list)  # Output: []
```

#List Slicing
* You can access a range of items in a list using slicing.
* list[start:stop:step]


```python
# Define a list
my_list = ['apple', 'banana', 'cherry', 'date', 'elderberry', 'fig', 'grape']

# Slicing examples
slice1 = my_list[1:4]        # Slice from index 1 to 3 (not including 4)
slice2 = my_list[:3]         # Slice from the beginning to index 2 (not including 3)
slice3 = my_list[4:]         # Slice from index 4 to the end of the list
slice4 = my_list[::2]        # Slice with a step of 2 (every other item)
slice5 = my_list[::]         # Slice with no start, stop, or step (returns the whole list)

# Print the results
print("Original List:", my_list)
print("Slice from index 1 to 3:", slice1)  # Output: ['banana', 'cherry', 'date']
print("Slice from the beginning to index 2:", slice2)  # Output: ['apple', 'banana', 'cherry']
print("Slice from index 4 to the end:", slice3)  # Output: ['elderberry', 'fig', 'grape']
print("Slice with a step of 2:", slice4)  # Output: ['apple', 'cherry', 'elderberry', 'grape']
print("Slice with no start, stop, or step:", slice5)  # Output: ['apple', 'banana', 'cherry', 'date', 'elderberry', 'fig', 'grape']
```

# Sorting and Other List Operations
* Using sort(): Sorts the list in ascending order (modifies the list in place).
* Using sorted(): Returns a new sorted list (does not modify the original list).

```python
# Define a list
my_list = [5, 2, 9, 1, 5, 6]

# Using sort() to sort the list in place
my_list.sort()
print("After sort():", my_list)  # Output: [1, 2, 5, 5, 6, 9]

# Define another list
another_list = [3, 1, 4, 1, 5, 9, 2, 6, 5]

# Using sorted() to get a new sorted list
sorted_list = sorted(another_list)
print("Original list:", another_list)  # Output: [3, 1, 4, 1, 5, 9, 2, 6, 5]
print("Sorted list:", sorted_list)     # Output: [1, 1, 2, 3, 4, 5, 5, 6, 9]

# Using sort() with the reverse parameter to sort in descending order
my_list.sort(reverse=True)
print("After sort(reverse=True):", my_list)  # Output: [9, 6, 5, 5, 2, 1]
```

List Functions
* min(): Returns the smallest item in the list.
* max(): Returns the largest item in the list.
* sum(): Returns the sum of all items in the list (items must be numeric).

```python
# Define a numeric list
numbers = [10, 20, 4, 45, 99]

# Using min() to find the smallest item
smallest = min(numbers)
print("Smallest item:", smallest)  # Output: 4

# Using max() to find the largest item
largest = max(numbers)
print("Largest item:", largest)  # Output: 99

# Using sum() to find the sum of all items
total_sum = sum(numbers)
print("Sum of items:", total_sum)  # Output: 178

# Example with an empty list for min() and max() with default
empty_list = []
smallest_empty = min(empty_list, default="No items")
largest_empty = max(empty_list, default="No items")
print("Smallest in empty list:", smallest_empty)  # Output: No items
print("Largest in empty list:", largest_empty)    # Output: No items
```

Tuples

* Tuples are a type of data structure in Python used to store multiple items in a single variable. They are:

1. Ordered: Items in a tuple have a defined order, which will not change.
Unchangeable (Immutable): Once created, items cannot be changed, added, or removed.
2. Allow Duplicate Values: Tuples can contain duplicate values because they are indexed.
Creating Tuples
3. You can create a tuple by placing a sequence of values inside parentheses, separated by commas. Tuples can also be created without parentheses.

```python
# Creating a tuple with parentheses
my_tuple1 = (1, 2, 3, 4, 5)
print("Tuple with parentheses:", my_tuple1)

# Creating a tuple without parentheses (using commas)
my_tuple2 = 6, 7, 8, 9, 10
print("Tuple without parentheses:", my_tuple2)

# Creating a tuple with a single item (requires a trailing comma)
single_item_tuple = (42,)
print("Single item tuple:", single_item_tuple)

# Creating an empty tuple
empty_tuple = ()
print("Empty tuple:", empty_tuple)

# Creating a tuple from a list
list_example = [1, 2, 3]
tuple_from_list = tuple(list_example)
print("Tuple from list:", tuple_from_list)
```

# Changing Tuple Values
*Tuples are immutable, meaning you cannot change their values directly. However, you can convert a tuple into a list, modify the list, and convert it back into a tuple.

```python
# Original tuple
original_tuple = (1, 2, 3, 4, 5)# Original tuple
original_tuple = (1, 2, 3, 4, 5)
print("Original Tuple:", original_tuple)

# Step 1: Convert the tuple to a list
temp_list = list(original_tuple)
print("List Converted from Tuple:", temp_list)

# Step 2: Modify the list
temp_list[2] = 99  # Change the third element from 3 to 99
temp_list.append(6)  # Add a new element 6
print("Modified List:", temp_list)

# Step 3: Convert the list back to a tuple
modified_tuple = tuple(temp_list)
print("Modified Tuple:", modified_tuple)
```

# set
Sets are another type of data structure in Python, used to store multiple items in a single variable. They are:

* Unordered: Sets do not maintain the order of items.
* No Duplicate Values: Sets automatically remove duplicate values.

# Creating Sets
You can create a set by placing items inside curly braces {}, separated by commas. You can also use the set() constructor.

```python
# Creating a set using curly braces
my_set = {1, 2, 3, 4, 5}
print("Set created with curly braces:", my_set)
# Creating a set using the set() constructor with a list
list_example = [1, 2, 3, 4, 5, 5, 6]
my_set_from_list = set(list_example)
print("Set created from list:", my_set_from_list)

# Creating a set using the set() constructor with a tuple
tuple_example = (1, 2, 3, 4, 5, 5, 6)
my_set_from_tuple = set(tuple_example)
print("Set created from tuple:", my_set_from_tuple)
```

Adding and Removing Items
* add(item): Adds an item to the set.
* pop(): Removes and returns an arbitrary item from the set.
* clear(): Removes all items from the set.

```python
# Create an initial set
my_set = {1, 2, 3}

# Add an item to the set
my_set.add(4)
print("Set after adding 4:", my_set)  # Output: {1, 2, 3, 4}

# Adding a duplicate item (no effect)
my_set.add(4)
print("Set after adding duplicate 4:", my_set)  # Output: {1, 2, 3, 4}
```

```python
# Create a set
my_set = {1, 2, 3, 4}

# Remove an arbitrary item
item_removed = my_set.pop()
print("Item removed:", item_removed)
print("Set after pop:", my_set)

# Removing from an empty set (uncomment to see the error)
# empty_set = set()
# empty_set.pop()  # Raises KeyError
```

```python
# Create a set
my_set = {1, 2, 3, 4, 5}

# Clear all items from the set
my_set.clear()
print("Set after clear:", my_set)  # Output: set()
```

# Python Dictionaries

## Introduction
Dictionaries in Python are used to store data values in key-value pairs. A dictionary is a collection that is ordered*, changeable, and does not allow duplicates.

## Dictionary Characteristics

- **Ordered**: As of Python 3.7 (and officially in Python 3.8), dictionaries maintain the order of items. This means that when iterating over a dictionary, items will be returned in the order they were added.
- **Changeable**: Dictionaries are mutable, meaning that their content can be changed after creation.
- **No Duplicates**: Keys in a dictionary must be unique. If you assign a value to an existing key, the old value will be overwritten.

## Dictionary Items

Dictionary items are presented in key-value pairs, and each pair is separated by a comma. Keys are unique identifiers used to access values within the dictionary.

### Example:


```python
thisdict = {
  'brand': 'Ford',
  'model': 'Mustang',
  'year': 1964
}
print(thisdict)
```

## Dictionary Length

To determine how many items a dictionary has, use the `len()` function:

### Example:


```python
thisdict = {
  'brand': 'Ford',
  'model': 'Mustang',
  'year': 1964
}
print(len(thisdict))
```

## Dictionary Items - Data Types

Values in a dictionary can be of any data type, and different data types can be mixed within the same dictionary.

### Example:


```python
thisdict = {
  'brand': 'Ford',
  'electric': False,
  'year': 1964,
  'colors': ['red', 'white', 'blue']
}
print(thisdict)
```

## Creating Dictionaries

### Using Curly Braces

The simplest way to create a dictionary is by using curly braces `{}`:

### Example:


```python
thisdict = {
  'name': 'John',
  'age': 36,
  'country': 'Norway'
}
print(thisdict)
```

### Using the `dict()` Constructor

You can also use the `dict()` constructor to create a dictionary:

### Example:


```python
thisdict = dict(name='John', age=36, country='Norway')
print(thisdict)
```

## Accessing Items

You can access the items of a dictionary by referring to its key name, inside square brackets. There is also a method called `get()` that will give you the same result:

### Example:


```python
thisdict = {
  'brand': 'Ford',
  'model': 'Mustang',
  'year': 1964
}

print(thisdict['brand'])  # Access by key name
print(thisdict.get('brand'))  # Access using get() method
```

## Get Keys, Values, and Items

You can retrieve all the keys, values, or key-value pairs from a dictionary using the `keys()`, `values()`, and `items()` methods, respectively.

### Example:


```python
thisdict = {
  'brand': 'Ford',
  'model': 'Mustang',
  'year': 1964
}

print(thisdict.keys())   # Get keys
print(thisdict.values()) # Get values
print(thisdict.items())  # Get items
```

## Check if Key Exists

To determine if a specified key is present in a dictionary, use the `in` keyword:

### Example:


```python
thisdict = {
  'brand': 'Ford',
  'model': 'Mustang',
  'year': 1964
}

print('model' in thisdict)  # Check if key exists
print('price' in thisdict)
```

## Adding or Changing Items

To add a new item to the dictionary, or to change the value of an existing item, you can use the assignment operator `=`:

### Example:


```python
thisdict = {
  'brand': 'Ford',
  'model': 'Mustang',
  'year': 1964
}

# Adding a new item
thisdict['color'] = 'red'
print(thisdict)

# Changing an existing item
thisdict['year'] = 2024
thisdict['color'] = 'blue'
print(thisdict)
```

## Removing Items

To remove an item, you can use the `pop()`, `popitem()`, or `del` methods:

### Example:


```python
thisdict = {
  'brand': 'Ford',
  'model': 'Mustang',
  'year': 1964
}

# Remove an item by key
thisdict.pop('year')
print(thisdict)

# Remove the last item
thisdict.popitem()
print(thisdict)

# Remove an item by key with del
del thisdict['model']
print(thisdict)
```

## Clear and Copy Dictionaries

You can empty a dictionary using the `clear()` method, and create a copy using the `copy()` method:

### Example:


```python
thisdict = {
  'brand': 'Ford',
  'model': 'Mustang',
  'year': 1964
}

# Copy the dictionary
mydict = thisdict.copy()
print(mydict)

# Clear the dictionary
thisdict.clear()
print(thisdict)

# Copy the dictionary again
mydict = thisdict.copy()
print(mydict)
```
