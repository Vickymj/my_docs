### String Methods

    * split()

```python
#Splits a string into a list based on a specified delimiter.

text = "Python is awesome"

words = text.split()
print(words)
```

* replace()

```python
#Replaces occurrences of a specified substring with another substring
text = "Python is awesome"
new_text = text.replace("awesome", "great")
print(new_text)
```

### upper()

```python
#Converts all characters in a string to uppercase.

text = "Python"
upper_text = text.upper()
print(upper_text)
```

* len()

```python
#Returns the length of an object (number of items in a container)

text = "Python"
length = len(text)
print(length)
```

* slicing

```python
#Slicing extracts a portion of a sequence (string, list, etc.) using the syntax sequence[start:stop:step].


text = "Python"
sliced_text = text[1:4]
print(sliced_text)
```

### Arithmetic operators:
* "+"	Addition
* '-'	Subtraction
*	Multiplication
* '/'	Division
* '%'	Modulus
* '**'	Exponentiation
* '//'	Floor division

```python
a, b = 10, 3
print(
    f"Addition: {a + b}, "
    f"Subtraction: {a - b}, "
    f"Multiplication: {a * b}, "
    f"Division: {a / b}, "
    f"Modulus: {a % b}, "
    f"Exponentiation: {a ** b}, "
    f"Floor Division: {a // b}"
)
```

### Comparison operators
* '=='	Equal
* '!='	Not equal
* '>'	Greater than
* '<'	Less than	x < y
* '>='	Greater than or equal to
* '<='	Less than or equal to

```python
x, y = 10, 5

print(
    f"x == y: {x == y}, "          # Equal
    f"x != y: {x != y}, "          # Not equal
    f"x > y: {x > y}, "            # Greater than
    f"x < y: {x < y}, "            # Less than
    f"x >= y: {x >= y}, "          # Greater than or equal to
    f"x <= y: {x <= y}"            # Less than or equal to
)
```

### Logical operators

```python
x, y = 10, 5

print(
    f"x > y and y > 0: {x > y and y > 0}, "  # True because both conditions are true
    f"x > y or y < 0: {x > y or y < 0}, "    # True because at least one condition is true
    f"not (x < y): {not (x < y)}"             # True because x < y is false
)
```
