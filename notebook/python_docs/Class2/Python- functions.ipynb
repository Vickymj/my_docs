## 1. Creating a Function
In Python, a function is defined using the `def` keyword. Functions are a way to group related statements together to perform a specific task.


`group of statement to perform a specifc task`

```python
# Example of creating a simple function
def greet():
    print("Hello, welcome to the Python functions tutorial!")
```

### 2. The `pass` Statement
The `pass` statement is used as a placeholder in functions, classes, loops, or conditional blocks. It does nothing and is used when the code is syntactically required but you don't want to write any code yet.

```python
# Example of using the pass statement in a function
def future_function():
    pass  # To be implemented later
```

### 3. Calling a Function
To call a function, use the function name followed by parentheses.

```python
# Example of calling the greet function
greet()
```

## 4. Function Arguments
### 4.1 Positional Arguments
Positional arguments are the most common type of argument. They are passed to the function in the same order in which they are defined.

```python
# Example of a function with positional arguments
def add(a, b):
    return a + b

result = add(5, 3)
print(result)  # Output: 8
```

### 4.2 Default Parameter Value
In Python, you can define default values for parameters in case no value is provided during the function call.

```python
# Example of a function with a default parameter value
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()  # Output: Hello, Guest!
greet("Alice")  # Output: Hello, Alice!
```

### 4.3 Keyword Arguments
Keyword arguments allow you to specify arguments by the parameter name during a function call.

```python
# Example of a function with keyword arguments
def describe_pet(pet_name, animal_type="dog"):
    print(f"I have a {animal_type} named {pet_name}.")

describe_pet(pet_name="Buddy")  # Output: I have a dog named Buddy.
describe_pet(animal_type="cat", pet_name="Whiskers")  # Output: I have a cat named Whiskers.
```

## 5. Arbitrary Arguments
### 5.1 Using `*args`
If you do not know how many arguments will be passed to your function, you can use `*args` to accept an arbitrary number of arguments.

```python
# Example of using *args
def greet_multiple(*names):
    for name in names:
        print(f"Hello, {name}!")

greet_multiple("Alice", "Bob", "Charlie")
```

### 5.2 Using `**kwargs`
`**kwargs` allows you to handle named arguments that you have not defined in advance.

```python
# Example of using **kwargs
def print_pet_info(**pet_info):
    for key, value in pet_info.items():
        print(f"{key}: {value}")

print_pet_info(name="Buddy", type="dog", age=5)
```

## 6. Advanced Example Combining `*args` and `**kwargs`
You can combine `*args` and `**kwargs` in a single function to accept both positional and named arguments.

```python
# Advanced example combining *args and **kwargs
def display_info(*args, **kwargs):
    print("Positional arguments:")
    for arg in args:
        print(arg)
    
    print("\nKeyword arguments:")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

display_info("Alice", "Bob", name="Charlie", age=30)
```
