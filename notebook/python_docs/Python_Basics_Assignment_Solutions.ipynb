### Problem 1: Variables and Operators

1. Create two variables `a` and `b` and assign them any integer values.
2. Perform the following operations and store the result in new variables:
   - Sum of `a` and `b`
   - Difference between `a` and `b`
   - Product of `a` and `b`
   - Division of `a` by `b`
   - Modulus of `a` by `b`
3. Print all the results.

```python
a = 10
b = 5

sum_result = a + b
diff_result = a - b
prod_result = a * b
div_result = a / b
mod_result = a % b

print('Sum:', sum_result)
print('Difference:', diff_result)
print('Product:', prod_result)
print('Division:', div_result)
print('Modulus:', mod_result)
```

### Problem 2: Working with Lists

1. Create a list of 5 different fruits.
2. Add another fruit to the list.
3. Remove the second fruit from the list.
4. Sort the list in alphabetical order.
5. Print the final list.

```python
fruits = ['Apple', 'Banana', 'Cherry', 'Date', 'Elderberry']
fruits.append('Fig')
fruits.pop(1)
fruits.sort()
print(fruits)
```

### Problem 5: Dictionaries

1. Create a dictionary with the following key-value pairs: name, age, and city.
2. Add a new key-value pair for country.
3. Update the age to a new value.
4. Print all the keys in the dictionary.
5. Print all the values in the dictionary.

```python
person = {'name': 'Alice', 'age': 25, 'city': 'New York'}
person['country'] = 'USA'
person['age'] = 26
print('Keys:', list(person.keys()))
print('Values:', list(person.values()))
```

### Problem 6: `if-else` Statements

1. Create a variable `num` and assign it any integer value.
2. Write an `if-else` statement to check if the number is even or odd.
3. Print whether the number is even or odd.

```python
num = 7
if num % 2 == 0:
    print(f'{num} is even')
else:
    print(f'{num} is odd')
```

### Problem 7: `for` Loops

1. Create a list of 5 integers.
2. Write a `for` loop to iterate through the list and print each number.
3. Modify the loop to print only the numbers greater than 10.

```python
numbers = [4, 12, 7, 15, 9]
for number in numbers:
    print(number)
for number in numbers:
    if number > 10:
        print(number)
```
