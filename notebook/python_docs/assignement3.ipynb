# Python Assignment3:

## 1. Lists and Slicing

**Q1:** Create a list `numbers` with elements [10, 20, 30, 40, 50]. Slice and print the first three elements.

**Q2:** Slice and print the last two elements of the `numbers` list.

**Q3:** Change the second element in the `numbers` list to 25 and print the updated list.

## 2. Loops and If Statements

**Q4:** Write a `for` loop that iterates over the `numbers` list and prints each element.

**Q5:** Write an `if` statement that checks if the number 40 is in the `numbers` list. If found, print "40 is in the list!"

**Q6:** Create a new list `even_numbers` containing only the even numbers from the `numbers` list using a `for` loop and `if` statement.

## 3. Functions

**Q7:** Write a function `square` that takes a number as input and returns its square.

**Q8:** Write a function `is_even` that checks if a number is even or odd. It should return True if the number is even, otherwise False.

**Q9:** Write a function `add_numbers` that takes two numbers as input and returns their sum.

## 4. Classes

**Q10:** Create a class `Student` with instance variables `name` and `age`. Create a method `introduce` that prints "Hi, my name is {name} and I am {age} years old."

**Q11:** Add a method `is_adult` to the `Student` class that returns `True` if the student is 18 or older, otherwise `False`.

## 5. Dictionaries

**Q12:** Create a dictionary `person` with the following key-value pairs: `'name': 'Alice', 'age': 25, 'city': 'New York'`. Access and print the value associated with the key `'age'`.

**Q13:** Add a new key-value pair `'email': 'alice@example.com'` to the `person` dictionary.

**Q14:** Write a loop that iterates over the `person` dictionary and prints each key and its corresponding value.

## 6. Combined Topics

**Q15:** Write a function `get_person_info` that takes a dictionary representing a person (with keys like `'name'`, `'age'`, `'city'`) as input, and returns a formatted string like "Name: {name}, Age: {age}, City: {city}".

**Q16:** Write a class `Book` that has instance variables `title`, `author`, and `price`. Create a method `details` that returns a formatted string with the book details.

**Q17:** Modify the `Book` class to include a method `apply_discount` that reduces the book's price by a given percentage.

## 7. Advanced Topics

**Q18:** Write a function `merge_dicts` that takes two dictionaries as input and returns a new dictionary containing all the key-value pairs from both dictionaries.

**Q19:** Create a list of dictionaries, where each dictionary represents a student with keys `'name'` and `'marks'`. Write a function `average_marks` that calculates and returns the average marks of all students.

**Q20:** Write a class `Car` with instance variables `brand` and `year`. Create a method `car_age` that calculates and returns the age of the car.
