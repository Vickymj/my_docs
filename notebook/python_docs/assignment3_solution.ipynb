## 4. Classes

**Q10:** Create a class `Student` with instance variables `name` and `age`. Create a method `introduce` that prints "Hi, my name is {name} and I am {age} years old."

```python
# Write your code here
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def introduce(self):
        print(f"Hi, my name is {self.name} and I am {self.age} years old.")

student1 = Student("John", 20)
student1.introduce()
```

**Q11:** Add a method `is_adult` to the `Student` class that returns `True` if the student is 18 or older, otherwise `False`.

```python
# Write your code here
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def introduce(self):
        print(f"Hi, my name is {self.name} and I am {self.age} years old.")
    
    def is_adult(self):
        return self.age >= 18

student1 = Student("John", 20)
print(student1.is_adult())  # True
```

**Q16:** Write a class `Book` that has instance variables `title`, `author`, and `price`. Create a method `details` that returns a formatted string with the book details.

```python
# Write your code here
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
    
    def details(self):
        return f"Title: {self.title}, Author: {self.author}, Price: {self.price}"

book = Book("Python 101", "John Doe", 29.99)
print(book.details())
```

**Q17:** Modify the `Book` class to include a method `apply_discount` that reduces the book's price by a given percentage.

```python
# Write your code here
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
    
    def details(self):
        return f"Title: {self.title}, Author: {self.author}, Price: {self.price}"
    
    def apply_discount(self, percentage):
        self.price -= self.price * (percentage / 100)

book = Book("Python 101", "John Doe", 29.99)
book.apply_discount(10)  # Applying 10% discount
print(book.details())
```

## 7. Advanced Topics

**Q18:** Write a function `merge_dicts` that takes two dictionaries as input and returns a new dictionary containing all the key-value pairs from both dictionaries.

```python
# Write your code here
def merge_dicts(dict1, dict2):
    merged = dict1.copy()
    merged.update(dict2)
    return merged

dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}
print(merge_dicts(dict1, dict2))
```

**Q19:** Create a list of dictionaries, where each dictionary represents a student with keys `'name'` and `'marks'`. Write a function `average_marks` that calculates and returns the average marks of all students.

```python
# Write your code here
students = [
    {'name': 'John', 'marks': 80},
    {'name': 'Alice', 'marks': 90},
    {'name': 'Bob', 'marks': 70}
]

def average_marks(students):
    total_marks = sum([student['marks'] for student in students])
    return total_marks / len(students)

print(average_marks(students))
```

**Q20:** Write a class `Car` with instance variables `brand` and `year`. Create a method `car_age` that calculates and returns the age of the car.

```python
# Write your code here
class Car:
    def __init__(self, brand, year):
        self.brand = brand
        self.year = year
    
    def car_age(self, current_year):
        return current_year - self.year

car = Car("Toyota", 2010)
print(car.car_age(2024))
```
