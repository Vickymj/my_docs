# Python OOP: Constructor, Class Variables, Instance Variables, and Inheritance

In this notebook, we will cover Object-Oriented Programming (OOP) concepts in Python:
- Constructor (`__init__` method)
- Instance and Class Variables
- Inheritance (Single, Multiple, and Multilevel)

## 1. Constructor in Python

A constructor is a special method used for initializing the object's state. The constructor is defined using the `__init__` method. The first argument `self` refers to the current object.

### Syntax:
```python
class ClassName:
    def __init__(self, parameters):
        # initialize object attributes
```

### Example:

````python
# 1. Constructor Example
class Car:
    # Constructor method to initialize instance variables
    def __init__(self, model, color, year):
        self.model = model  # instance variable
        self.color = color  # instance variable
        self.year = year    # instance variable
        
    def display_car_details(self):
        print(f"Model: {self.model}, Color: {self.color}, Year: {self.year}")

# Creating a Car object
car1 = Car('Toyota Camry', 'Red', 2022)

# Display car details
car1.display_car_details()
````

## 2. Class Variables vs Instance Variables

- **Instance Variables:** These are defined within the constructor (`__init__`) and are unique to each object (instance).
- **Class Variables:** These are defined outside the constructor and shared by all instances of the class.

### Example:

````python
# 2. Class Variables vs Instance Variables Example
class Car:
    # Class variable (shared across all instances)
    car_type = 'Sedan'
    
    def __init__(self, model, color, year):
        self.model = model  # instance variable
        self.color = color  # instance variable
        self.year = year    # instance variable
        
    def display_car_details(self):
        print(f"Type: {Car.car_type}, Model: {self.model}, Color: {self.color}, Year: {self.year}")

# Creating multiple Car objects
car1 = Car('Honda Accord', 'Blue', 2020)
car2 = Car('Tesla Model 3', 'White', 2021)

# Display car details
car1.display_car_details()
car2.display_car_details()
````

## 3. Inheritance in Python
Inheritance is a mechanism where a new class (child) inherits the properties and behaviors of another class (parent). It allows reusability of code and extends the functionality of the parent class.

### 3.1 Single Inheritance
In single inheritance, a child class inherits from one parent class.

### Example:

````python
# 3.1 Single Inheritance Example
class Animal:
    def __init__(self, species):
        self.species = species
    
    def speak(self):
        print(f"The {self.species} makes a sound.")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__('Dog')  # Calling the parent class constructor
        self.name = name
        self.breed = breed
    
    def speak(self):
        print(f"{self.name} the {self.breed} barks.")

# Creating a Dog object (inheriting from Animal)
dog = Dog('Buddy', 'Golden Retriever')
dog.speak()
````

### 3.2 Multiple Inheritance
In multiple inheritance, a child class can inherit from more than one parent class. This allows the child class to inherit attributes and methods from multiple classes.

### Example:

````python
# 3.2 Multiple Inheritance Example
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower
    
    def start(self):
        print(f"Engine with {self.horsepower} horsepower starts.")

class Body:
    def __init__(self, body_type):
        self.body_type = body_type
    
    def shape(self):
        print(f"Car body type is {self.body_type}.")

class Car(Engine, Body):
    def __init__(self, model, color, year, horsepower, body_type):
        Engine.__init__(self, horsepower)
        Body.__init__(self, body_type)
        self.model = model
        self.color = color
        self.year = year
    
    def display_details(self):
        print(f"Model: {self.model}, Color: {self.color}, Year: {self.year}")

# Creating a Car object (inherits from Engine and Body)
car = Car('Tesla Model S', 'Red', 2022, 670, 'Sedan')

# Display car details
car.start()
car.shape()
car.display_details()
````

### 3.3 Multilevel Inheritance
In multilevel inheritance, a child class inherits from a parent class, which itself is a child of another parent class. This forms a chain of inheritance.

### Example:

````python
# 3.3 Multilevel Inheritance Example
class Vehicle:
    def __init__(self, make):
        self.make = make
    
    def start(self):
        print(f"{self.make} vehicle is starting.")

class Car(Vehicle):
    def __init__(self, make, model):
        super().__init__(make)
        self.model = model
    
    def drive(self):
        print(f"{self.make} {self.model} is driving.")

class ElectricCar(Car):
    def __init__(self, make, model, battery_capacity):
        super().__init__(make, model)
        self.battery_capacity = battery_capacity
    
    def charge(self):
        print(f"{self.make} {self.model} is charging with {self.battery_capacity} kWh battery.")

# Creating an ElectricCar object
ev = ElectricCar('Tesla', 'Model 3', 75)
ev.start()
ev.drive()
ev.charge()
````

## Summary
- The constructor (`__init__`) is used to initialize object attributes.
- Class variables are shared across all instances, while instance variables are unique to each object.
- Inheritance allows us to reuse code and create a hierarchy of classes. In Python, we can use single, multiple, and multilevel inheritance to create more flexible designs.
